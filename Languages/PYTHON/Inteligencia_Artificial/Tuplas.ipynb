```python
# indexado
print("valor de [0] ", tupla1[0])
print("valor de [1] ", tupla1[1])
print("valor de [2] ", tupla1[2])
```

```python
# indexado negativo
print("valor de [-1] ", tupla1[-1])
print("valor de [-2] ", tupla1[-2])
print("valor de [-3] ", tupla1[-3])
```

```python
# metodo soting
ratings = (0,-1,8,7,6,4)
print("tupla sin organizar ", ratings)
tupla_organizada = sorted(ratings)
print("tupla organizada ", tupla_organizada)
```
