# Realiza un listado de 0 a 100 mostrando solo los numeros primos

```python
cont = 0
lista_primos = []
for i in range(100): 
    for j in range(1,i+1):
        residuo = i%j
        if residuo == 0: 
            cont = cont + 1
    if cont ==2:
        lista_primos.append(i)

    cont = 0

lista_primos
```

### crear una funcion que calcule un porcentaje dado de un valor

```python
def porcentaje (num1, num2):
    """
    Dado 2 valores el primero porcentaje y el segundo valor.
    Calcula el porcentaje especificado.

    argumento
    num1: es el porcentaje de 0 - 100
    num2: es el valor
    """
    p = num2 * num1/100

    return p
```

```python
help(porcentaje)
```

```python
# Calcular el porcentaje  20% del 100.000
p = porcentaje(20, 100000)
p
```

```python
# Calcular el porcentaje  20% del 150.000
p = porcentaje(20, 150000)
p
```
