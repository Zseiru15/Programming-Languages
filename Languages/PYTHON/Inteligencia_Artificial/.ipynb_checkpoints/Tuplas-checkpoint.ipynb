# Tuplas

```python
# ()
tupla1 = ("Disco", 10, 1.5)
tupla1
```

```python
type(tupla1)
```

```python
a = 1,2,4,5,7
type(a)
```

```python
# indexado
print("valor de [0] ", tupla1[0])
print("valor de [1] ", tupla1[1])
print("valor de [2] ", tupla1[2])
```

```python
# indexado negativo
print("valor de [-1] ", tupla1[-1])
print("valor de [-2] ", tupla1[-2])
print("valor de [-3] ", tupla1[-3])
```

# Concatenar

```python
# tupla 2
tupla2 = tupla1 + ("hard rock", 54)
tupla2
```
