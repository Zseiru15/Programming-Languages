### Ejercicio 3

Las edades de **5** estudiantes del turno mañana.

### Ejercicio 4 

Ingresar una oración que pueden tener letras tanto en *mayúsculas* como *minúsculas*. Contar la cantidad de vocales. Crear un segundo string con toda la oración en minúsculas para que sea más fácil disponer la condición que verifica que es una vocal.

### Ejercicio 5

Crear y cargar en un lista los nombres de 5 países y en otra lista paralela la cantidad de habitantes del mismo. Ordenar alfabéticamente e imprimir los resultados. Por último ordenar con respecto a la cantidad de habitantes (de mayor a menor) e imprimir nuevamente.

```python
### tu código aquí ###

paises = ['Argentina', 'México', 'Colombia', 'España', 'Perú']
habitantes = [46000000, 128000000, 52000000, 48000000, 34000000]

lista_alfabetica = sorted(zip(paises, habitantes))
print('Ordenado alfabéticamente')
for pais, hab in lista_alfabetica:
    print(f'País: {pais:<10} | Habitantes: {hab:,}')

lista_habitantes = sorted(zip(paises, habitantes), key=lambda x: x[1], reverse=True)
print('\nOrdenado por habitantes (Mayor a Menor)')
for pais, hab in lista_habitantes:
    print(f'País: {pais:<10} | Habitantes: {hab:,}')
```

### Ejercicio 7
Confeccionar una función que reciba una serie de edades y me retorne la cantidad que son mayores o iguales a 18 (como mínimo se envía un entero a la función

```python
### tu código aquí ###

def contar_mayores_edad(*edades):
    cantidad = 0
    for edad in edades:
        if edad >= 18:
            cantidad += 1
    return cantidad

resultado = contar_mayores_edad(12, 18, 25, 40, 15)
print(f"Cantidad de mayores o iguales a 18: {resultado}")
```

### Ejercicio 8 

Almacenar en una lista 5 empleados, cada elemento de la lista es una sub lista con el nombre del empleado junto a sus últimos tres sueldos (estos tres valores en una tupla)

El programa debe tener las siguientes funciones:

1) Carga de los nombres de empleados y sus últimos tres sueldos.

2) Imprimir el monto total cobrado por cada empleado.

3) Imprimir los nombres de empleados que tuvieron un ingreso trimestral mayor a 10000 en los últimos tres meses.

Tener en cuenta que la estructura de datos si se carga por asignación debería ser similar a:

<code>empleados = [["juan",(2000,3000,4233)] , ["ana",(3444,1000,5333)] ,  etc.   ]</code>

```python
### tu código aquí ###

def cargar_empleados():
    '''Carga los nombres de 5 empleados y sus últimos tres sueldos en una tupla.'''
    empleados = []
    for i in range(5):
        print(f'\nEmpleado {i + 1}:')
        nombre = input('Ingrese el nombre del empleado: ')
        sueldo1 = float(input('Ingrese el sueldo del mes 1: '))
        sueldo2 = float(input('Ingrese el sueldo del mes 2: '))
        sueldo3 = float(input('Ingrese el sueldo del mes 3: '))
        
        tupla_sueldos = (sueldo1, sueldo2, sueldo3)
        empleados.append([nombre, tupla_sueldos])
        
    return empleados

def imprimir_total_cobrado(empleados):
    '''Calcula e imprime el monto total cobrado por cada empleado.'''
    print('\nMONTO TOTAL COBRADO POR CADA EMPLEADO')
    for empleado in empleados:
        nombre = empleado[0]
        sueldos = empleado[1]
        total = sum(sueldos)
        print(f"Empleado: {nombre} | Total cobrado: ${total:.2f}")

def imprimir_ingreso_mayor_a_10000(empleados):
    '''Imprime los nombres de los empleados con un ingreso trimestral mayor a 10000.'''
    print('\nEMPLEADOS CON INGRESO TRIMESTRAL MAYOR A 10000')
    encontrados = False
    for empleado in empleados:
        nombre = empleado[0]
        sueldos = empleado[1]
        total = sum(sueldos)
        if total > 10000:
            print(f"Empleado: {nombre} (Total: ${total:.2f})")
            encontrados = True
    
    if not encontrados:
        print('Ningún empleado supera los 10000 en el trimestre.')

def main():
    lista_empleados = cargar_empleados()
    # lista_empleados = [
    #     ["juan", (2000, 3000, 4233)],
    #     ["ana", (3444, 5000, 5333)],
    #     ["carlos", (1500, 2000, 1800)],
    #     ["lucia", (4000, 4000, 3000)],
    #     ["pedro", (1000, 1200, 900)]
    # ]
    
    imprimir_total_cobrado(lista_empleados)
    imprimir_ingreso_mayor_a_10000(lista_empleados)

if __name__ == "__main__":
    main()
```

### Ejercicio 9

Crear un diccionario en Python que defina como clave el número de documento de una persona y como valor un string con su nombre.

Desarrollar las siguientes funciones:

1) Cargar por teclado los datos de 4 personas.
2) Listado completo del diccionario.
3) Consulta del nombre de una persona ingresando su número de documento.

```python
### tu código aquí ###

def cargar_personas():
    '''Carga por teclado los datos de 4 personas en un diccionario.'''
    diccionario = {}
    print('Carga de Datos (4 personas)')
    for i in range(4):
        documento = input(f'Ingrese el número de documento de la persona {i+1}: ')
        nombre = input(f'Ingrese el nombre de la persona {i+1}: ')
        diccionario[documento] = nombre
        print('-' * 30)
    return diccionario

def listar_diccionario(diccionario):
    '''Muestra el listado completo del diccionario.'''
    print('\nListado Completo de Personas')
    if not diccionario:
        print('El diccionario está vacío.')
    else:
        for documento, nombre in diccionario.items():
            print(f'Documento: {documento} - Nombre: {nombre}')

def consultar_por_documento(diccionario):
    '''Consulta el nombre de una persona ingresando su número de documento.'''
    print('\nConsulta de Persona')
    busqueda = input('Ingrese el número de documento que desea consultar: ')
    if busqueda in diccionario:
        print(f'El nombre asignado a ese documento es: {diccionario[busqueda]}')
    else:
        print('El número de documento no se encuentra en el diccionario')

personas = cargar_personas()

listar_diccionario(personas)

consultar_por_documento(personas)
```

### Ejercicio 10

Crear un diccionario en Python para almacenar los datos de empleados de una empresa. La clave será su número de legajo y en su valor almacenar una lista con el nombre, profesión y sueldo.

Desarrollar las siguientes funciones:

1) Carga de datos de empleados.
2) Permitir modificar el sueldo de un empleado. Ingresamos su número de legajo para buscarlo.
3) Mostrar todos los datos de empleados que tienen una profesión de "analista de sistemas"

```python
### tu código aquí ###

def cargar_empleados():
    '''Permite ingresar los datos de los empleados por teclado.'''
    empleados = {}
    print('Carga de Datos de Empleados')
    while True:
        legajo = input('Ingrese el número de legajo (o "fin" para terminar): ').strip()
        if legajo.lower() == 'fin':
            break

        if legajo in empleados:
            print('Este número de legajo ya existe. Intente con otro.')
            continue
            
        nombre = input('Nombre: ').strip()
        profesion = input('Profesión: ').strip().lower()
        
        try:
            sueldo = float(input('Sueldo: '))
        except ValueError:
            print('Sueldo no válido. Se asignará 0.0')
            sueldo = 0.0

        empleados[legajo] = [nombre, profesion, sueldo]
        print('Empleado cargado con éxito.\n')
        
    return empleados


def modificar_sueldo(empleados):
    '''Busca un empleado por legajo y modifica su sueldo.'''
    print('\nModificar Sueldo de Empleado')
    legajo = input('Ingrese el número de legajo del empleado a modificar: ').strip()
    
    if legajo in empleados:
        print(f'Empleado encontrado: {empleados[legajo][0]} | Sueldo actual: ${empleados[legajo][2]}')
        try:
            nuevo_sueldo = float(input('Ingrese el nuevo sueldo: '))
            empleados[legajo][2] = nuevo_sueldo
            print('Sueldo actualizado correctamente.')
        except ValueError:
            print('Monto no válido. No se realizaron cambios.')
    else:
        print('El número de legajo no existe en la base de datos.')


def mostrar_analistas(empleados):
    '''Muestra todos los empleados con la profesión de 'analista de sistemas'.'''
    print('\nEmpleados con profesión "Analista de Sistemas" ---')
    encontrado = False
    
    for legajo, datos in empleados.items():
        nombre, profesion, sueldo = datos
        if profesion == 'analista de sistemas':
            print(f'Legajo: {legajo} | Nombre: {nombre} | Sueldo: ${sueldo}')
            encontrado = True
            
    if not encontrado:
        print('No se encontraron empleados con la profesión de "analista de sistemas".')

if __name__ == "__main__":
    mis_empleados = cargar_empleados()
    if mis_empleados:
        modificar_sueldo(mis_empleados)
        mostrar_analistas(mis_empleados)
    else:
        print('No se registraron empleados para trabajar.')
```
