```python
A = [1, 3, 4, 5, 6]
B = A
print(B)
A[2] = 7
print(B)
```

```python
A = [1, 3, 4, 5, 6]
B = A.copy()
print(B)
A[2] = 7
print(B)
```

# Ejercicio 3 Mutabilidad, Edicion y Eliminacion

```python
A = ["Disco", 10, 1.2]
```

```python
print("Original sin cambio ", A)
A[0] = "Hard Rock"
print("Alterada con cambio ", A)
```

```python
print("Original sin eliminar ", A)
del(A[0])
print("Alterada con eliminacion ", A)
```

```python
print("Resultado final ", A)
```

# Diccionarios

```python
# {Key:Valor}

diccionario1 = {
    'nombre': 'Katerine',
    'celular': 324879788,
    (1,2): 2.232,
    5: 'edad',
    5.78: [5.47, 78.9]
}
```

```python
diccionario1
```

```python
# Las llaves pueden ser enteros, string, float y tuplas.
diccionario1['nombre']
```

```python
print(diccionario1['nombre'])
print(diccionario1['celular'])
print(diccionario1[(1,2)])
print(diccionario1[5])
print(diccionario1[5.78])
```

```python
diccionario1['Diccion'] = 'calle 47# 78-125'
diccionario1
```

```python
type(diccionario1)
```

```python
# Retornar las claves sin los valores
list(diccionario1.keys())
```

```python
# Retornar los valores de las llaves
diccionario1.values()
```

```python
# La palabra reservada in
'nombre' in diccionario1
```
